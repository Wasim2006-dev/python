#top ten district by population

```python
top10=df.nlargest(10, "Population")
print(top10)
```

```python
df.head(10)
```

```python
plt.figure(figsize=(12, 6))
plt.bar(top10["District_name"], top10["Population"])
plt.xticks(rotation=45, ha="right")
plt.title("Top 10 Districts by Population")
plt.xlabel("District")
plt.ylabel("Population")
plt.tight_layout()
plt.show()
```

#calculate the male and female ratio

```python
df["male_female_ratio"] = (df["Male"] / df["Female"]).round(2)
print(df)
```

```python
top10 = df.nlargest(10, "Population")

plt.figure(figsize=(12,6))
plt.bar(top10["District_name"], top10["male_female_ratio"])
plt.title("Top 10 Districts by Male-Female Ratio")
plt.xlabel("District")
plt.ylabel("Male : Female Ratio")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#average litracy by state

```python
state_literacy=(df.groupby("State_name")["Literate"].mean().sort_values(ascending=False))
```

top10

```python
top10=state_literacy.head(10)
print(top10)
```

```python
plt.figure(figsize=(10,6))
top10.plot(kind="bar")
plt.title("Top 10 Most Literate States")
plt.xlabel("State")
plt.ylabel("Average Literacy Rate (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#religion by state

```python
religion_state = df.groupby("State_name")[[
    "Hindus",
    "Muslims",
    "Christians",
    "Sikhs",
    "Buddhists",
    "Jains"
]].sum()
print(religion_state)
religion_state.plot(kind="bar", stacked=True, figsize=(14,7))

plt.title("Religion-wise Population by State")
plt.xlabel("State")
plt.ylabel("Population")
plt.xticks(rotation=90)
plt.legend(title="Religion")
plt.tight_layout()
plt.show()
```

# age group by stae

```python
age_group = df.groupby("State_name")[[
    "Age_Group_0_29",
    "Age_Group_30_49",
    "Age_Group_50"
]].sum()
print(age_group)
age_group.plot(kind="bar", stacked=True, figsize=(14,7))

plt.title("Age Group Distribution by State")
plt.xlabel("State")
plt.ylabel("Population")
plt.xticks(rotation=90)
plt.legend(title="Age Group")
plt.tight_layout()
plt.show()
```
